```python
class Car:
    def __init__(self, model, make):
        self.model = model
        self.make = make

    def start(self):
        print(f"{self.make} {self.model} => Car started......")
    
    def stop(self):
        print(f"{self.make} {self.model} => Car stopped......")

    def __str__(self):
        """Inherited and changes the behavior of the string representation of the object."""
        return f"make = {self.make}, model = {self.model}"
```

```python
class Account:
    def __init__(self, number, holder, balance):
        self.number = number
        self.holder = holder
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        
    def withdraw(self, amount):
        self.balance -= amount

    def transfer(self, to_account, amount):
        self.withdraw(amount)
        to_account.deposit(amount)
            
    def __str__(self):
        return f"Account Number: {self.number}, Account Holder: {self.holder}, Balance: {self.balance}"
```

```python

jay_account = Account("1", "Jay A", 0)
viru_account = Account("2", "Viru B", 0)
thakur_account = Account("3", "Thakur C", 10000000)
```

```python
print(f"thakur_account balance: {thakur_account.balance}" )
```

```python
def print_balances():
    print(f"jay_account balance: {jay_account.balance}")
    print(f"viru_account balance: {viru_account.balance}")
    print(f"thakur_account balance: {thakur_account.balance}")
```

```python
print_balances()
```

```python
thakur_account.transfer(jay_account, 10000)
thakur_account.transfer(viru_account, 10000)
```

```python
print_balances()
```

```python
class TrainTicket:
    def __init__(self, ticket_id, train_name, journey_date, seat_number, price, from_station, to_station, passenger_name):
        self.ticket_id = ticket_id
        self.train_name = train_name
        self.journey_date = journey_date
        self.seat_number = seat_number
        self.price = price
        self.from_station = from_station
        self.to_station = to_station
        self.passenger_name = passenger_name

    def train_route(self):
        return f"{self.from_station} to {self.to_station}"

    def ticket_details(self):
        return f"Ticket ID: {self.ticket_id}, Train: {self.train_name}, Date: {self.journey_date}, Seat: {self.seat_number}, Price: {self.price}, Route: {self.train_route()}"

    def customer_details(self):
        return f"Passenger Name: {self.passenger_name}", f"Route: {self.train_route()}", f"Seat: {self.seat_number}"
```

```python
class Ticket:
    # class member
    total_tickets_sold = 0

    def __init__(self, price = 0):
        self.price = price
        self.increment_count()

    #instance method
    def print_ticket(self):
        print(f"Printing ticket for sale of {self.price}")

    @classmethod
    def increment_count(cls):
        cls.total_tickets_sold += 1

    #This is a utility method
    @staticmethod
    def is_price_valid(amount):
        return amount >= 50
```

```python
t1 = Ticket(100)
t2 = Ticket(200)
t3 = Ticket(300)
```

```python
Ticket.total_tickets_sold
```

```python
Ticket.is_price_valid(40)
```
