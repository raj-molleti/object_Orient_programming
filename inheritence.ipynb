```python
class Account:
    def __init__(self, number, balance):
        self.number = number
        self.balance = balance

    def withdraw(self, amount):
        self.balance -= amount

    def deposit(self, amount):
        self.balance += amount  
```

```python
# SavingsAccount inherits from Account
class SavingsAccount(Account):
    pass
```

```python
#current account derived from Account
class CurrentAccount(Account):
    pass
```

```python
c1 = CurrentAccount(number=1, balance=50000)
```

```python
s1 = SavingsAccount(number=2, balance=30000)
```

```python
s1.balance
```

```python
c1.balance
```

###saving account
* your are not allowed to withdraw or deposit more than 50000 in SavingsAccount

```python
# SavingsAccount rules
# You are not allowed to withdraw or deposit more than 50000 in SavingsAccount
class SavingsAccount(Account):

    def withdraw(self, amount):
        if amount >= 50000:
            print("You are not allowed to withdraw more than 50000 in SavingsAccount")
        else:
            super().withdraw(amount)

    def deposit(self, amount):
        if amount >= 50000:
            print("You are not allowed to deposit more than 50000 in SavingsAccount")
        else:
            # call base class method
            super().deposit(amount)
```

```python
s1 = SavingsAccount(1, 100000)
```

```python
s1.withdraw(100000)
```

```python
s1.deposit(100000)
```

```python
s1.withdraw(10000)
s1.balance
```

```python
class Account(Account):
    def __init__(self, number, balance):
        #private
        self._number = number
        #protected
        self.__balance = balance
    
    def deposit(self, amount):
        self.__balance += amount

    def get_balance(self):
            return self.__balance
```
