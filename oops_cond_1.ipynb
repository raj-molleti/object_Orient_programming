```python
class Shape:
    def area(self):
        pass
```

```python
class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14 * self.radius ** 2
```

```python
class Rectangle(Shape):
    def __init__(self, width, length):
        self.width = width
        self.length = length

    def area(self):
        return self.width * self.length
```

```python
c1 = Circle(5)  
r1 = Rectangle(4, 6)
```

```python
#membership operators
isinstance(c1, Circle)
```

```python
isinstance(c1, Shape)
```

```python
issubclass(Circle, Shape)
```

```python
def calculate_area(shape):
    return shape.area()
```

```python
calculate_area(c1)
```

```python
calculate_area(r1)
```

```python
#Duck Typing
class Animal:
    def make_sound(self):
        print("....Roaring....")

class Vechile:
    def make_sound(self):
        print("....Vroom Vroom....")

def make_it_sound(sample):
    sample.make_sound()
```

```python
a = Animal()
v = Vechile()
make_it_sound(a)
make_it_sound(v)
```

```python
def is_even(number):
    if not isinstance(number, int):
        return "Not a number"

    if number == 0:
        return False
    
    return number % 2 == 0
```

```python
is_even(4)
```

```python
is_even(-2)
```

```python
is_even(0)
```

```python
is_even("2")
```

```python
from abc import ABC, abstractmethod
class Animal(ABC):

    @abstractmethod
    def sound(self):
        pass
```

```python
class Dog(Animal):
    def sound(self):
        return "Barking...."
```

```python
class Cat(Animal):
    def sound(self):
        return "Meowing...."
```
