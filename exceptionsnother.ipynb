```python
def div(a, b):
    try:
        return a / b
    except:
        print("An error occurred")
    return None
```

```python
div(10, 5)
```

```python
div(10, 0)
```

```python
import traceback

def div_1(a, b):
    result = None
    try:
        result = a / b
    except ZeroDivisionError as ze:
        print(f"Division by zero is not allowed: {ze}")
        traceback.print_exc()
    else:
        print("No exception occurred")
    finally:
        print("Execution of try-except block is complete")
    return result
```

```python
div_1(10,0)
```
